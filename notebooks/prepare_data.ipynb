### Создаем копию для подготовки

```python
df_prepared = df_moscow.copy()
```

## Добавляем год и месяц

```python
df_prepared['year'] = df_prepared['datetime'].dt.year
df_prepared['month'] = df_prepared['datetime'].dt.month

df_prepared.head()
```

## Добавляем сезон

```python
pd.set_option('display.max_columns', None)
```

```python
seasons = {
    1: 'Зима',
    2: 'Зима',
    3: 'Весна',
    4: 'Весна',
    5: 'Весна',
    6: 'Лето',
    7: 'Лето',
    8: 'Лето',
    9: 'Осень',
    10: 'Осень',
    11: 'Осень',
    12: 'Зима'
}

df_prepared['season'] = df_prepared['month'].map(seasons)
df_prepared.head()
```

## Добавляем день недели

```python
df_prepared['day_of_week_num'] = df_prepared['datetime'].dt.dayofweek + 1

weekdays = {
    1: 'Понедельник',
    2: 'Вторник',
    3: 'Среда',
    4: 'Четверг',
    5: 'Пятница',
    6: 'Суббота',
    7: 'Воскресенье'
}

df_prepared['day_of_week'] = df_prepared['day_of_week_num'].map(weekdays)
df_prepared.head()
```

## Добавляем час

```python
df_prepared['hour'] = df_prepared['datetime'].dt.hour
df_prepared.head()
```

## Добавляем время суток

```python
df_prepared['time_of_day'] = ''

df_prepared.loc[df_prepared['hour'].between(0, 5), 'time_of_day'] = '00:00-06:00'
df_prepared.loc[df_prepared['hour'].between(6, 11), 'time_of_day'] = '06:00-12:00'
df_prepared.loc[df_prepared['hour'].between(12, 17), 'time_of_day'] = '12:00-18:00'
df_prepared.loc[df_prepared['hour'].between(18, 23), 'time_of_day'] = '18:00-24:00'

df_prepared.head()
```

```python
df_prepared['time_of_day'].value_counts()
```

## Подготавливаем освещение

```python
light_groups = {
    'Светлое время суток': 'Дневной свет',
    'Сумерки': 'Сумерки',
    'В темное время суток, освещение включено': 'Темное время суток, освещение включено',
    'В темное время суток, освещение отсутствует': 'Темное время суток, освещение отсутствует',
    'В темное время суток, освещение не включено': 'Темное время суток, освещение не включено',
    'Не установлено': 'Не установлено'
}

df_prepared['light_group'] = df_prepared['light'].map(light_groups)
df_prepared['light_group'].value_counts()
```

## Создаем группы погодных условий

```python
weather_groups = {
    'Ясно': 'Ясно',
    'Температура выше +30С, Ясно': 'Ясно',
    'Пасмурно': 'Пасмурно',
    'Дождь': 'Дождь',
    'Дождь, Пасмурно': 'Дождь',
    'Дождь, Ясно': 'Дождь',
    'Снегопад': 'Снегопад',
    'Пасмурно, Снегопад': 'Снегопад',
    'Дождь, Снегопад': 'Дождь со снегом',
    'Туман': 'Туман',
    'Пасмурно, Туман': 'Туман',
    'Дождь, Туман': 'Дождь с туманом',
    'Метель': 'Метель',
    'Метель, Снегопад': 'Метель',
    'Метель, Пасмурно': 'Метель'
}

df_prepared['weather_group'] = df_prepared['weather'].map(weather_groups)
df_prepared['weather_group'].value_counts()
```

## Подготовка территории

```python
df_prepared['district'] = df_prepared['county'].fillna('Район не указан')

print('ДТП с указанной территорией:', (df_prepared['county'] != 'Район не указан').sum())
print('ДТП без указанной территории:', (df_prepared['district'] == 'Район не указан').sum())
```

## Добавляем признак наличия погибших

```python
df_prepared['has_dead'] = (df_prepared['dead_count'] > 0).astype(int)

print('ДТП с погибшими:', df_prepared['has_dead'].sum())
df_prepared[['dead_count', 'has_dead']].head(10)
```

## Проверка согласованности освещения с датой и временем

```python
winter_daylight = df_prepared[
    df_prepared['datetime'].dt.month.isin([12, 1, 2])
    & df_prepared['datetime'].dt.hour.between(0, 5)
    & (df_prepared['light'] == 'Светлое время суток')
]

print('Подозрительных записей:', len(winter_daylight))

print('Доля от всей выборки, %:', round(len(winter_daylight) / len(df_prepared) * 100, 3))

winter_daylight[['id', 'datetime', 'light', 'district']].head(20)
```

```python
summer_darkness = df_prepared[
    df_prepared['datetime'].dt.month.isin([6, 7, 8])
    & df_prepared['datetime'].dt.hour.between(12, 17)
    & df_prepared['light'].str.startswith('В темное время суток')
]

print('Подозрительных записей:', len(summer_darkness))

print('Доля от всей выборки, %:', round(len(summer_darkness) / len(df_prepared) * 100, 3))

summer_darkness[['id', 'datetime', 'light', 'district']].head(20)
```

```python
suspect_light_count = len(winter_daylight) + len(summer_darkness)

print('Всего подозрительных записей по освещению:', suspect_light_count)

print('Доля от всей выборки, %:', round(suspect_light_count / len(df_prepared) * 100, 3))
```

При сопоставлении условий освещения с датой и временем обнаружены подозрительные сочетания: светлое время суток зимой с 00:00 до 06:00 и темное время суток летом с 12:00 до 18:00. По этим двум правилам выявлены 124 записи, или около 0,184% выборки. Проверка предполагает, что время происшествий указано по московскому местному времени.

По имеющимся данным невозможно установить, какое исходное поле ошибочно: дата, время или освещение. Поэтому исправление значений и исключение записей не выполнялись.

Выявленная несогласованность рассматривается как ограничение качества датасета. Результаты анализа освещения характеризуют категории, указанные в источнике. Проверка охватывает только выбранные сочетания зимой и летом и не определяет полную долю ошибочных записей.
